Diese Funktion berechnet die Collatz-Folge für eine gegebene Zahl n.
Dabei wird folgende Formel verwendet:
$$
f(n)=
\begin{cases}
\frac{n}{2} & \text{falls } n \text{ gerade} \\
3n+1 & \text{falls } n \text{ ungerade}
\end{cases}
$$
#### Variablen
- result_list -> die fertige Liste der collatz folge
- current_num -> die aktuelle Num

```python
def collatz(n):
    result_list = [n]
    current_num = n

    while current_num != 1:
        if current_num % 2 == 0:
            current_num //= 2
            result_list.append(current_num)
        elif current_num % 2 != 0:
            current_num = (current_num * 3) + 1
            result_list.append(current_num)

    return result_list
```

#### Teil 1 - Ausprobieren der collatz function mit 3 verschiedenen Zahlen

```python
a, b, c = 134, 27, 19

print(collatz(a))
print(collatz(b))
print(collatz(c))
```

#### Teil 2 - Rekordsuche

##### Alle Startzahlen von 1 - 10_000 untersuchen und folgende dinge herausfinden
- Startzahl mit der Längsten folge
- Startzahl, die das höchste Maximum erreicht
- Wie viele Zahlen mehr als 100 Steps brauchen

```python
longest_sequence = [0, 0]
highest_max = [0, 0]
more_than_100_steps_count = 0

for i in range(1, 10_001):
    collatz_res = collatz(i)

    if len(collatz_res) > longest_sequence[1]:
        longest_sequence = [i, len(collatz_res)]

    if max(collatz_res) > highest_max[1]:
        highest_max = [i, max(collatz_res)]

    if (len(collatz_res) - 1) > 100:
        more_than_100_steps_count += 1

print(f"Längste Folge bei {longest_sequence[0]} mit {longest_sequence[1]} Folgengliedern")
print(f"Größte Zahl in der folge bei {highest_max[0]} mit {highest_max[1]}")
print(f"Zahlen mit mehr als 100 Folgengliedern: {more_than_100_steps_count}")
```
