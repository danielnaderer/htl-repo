`*` entspricht Varargs in Java

```python
def sum(*values):
    sum = 0
    for item in values:
        sum += item
    return sum

sum(1, 2, 3)
```

* `list[index]` liefert ein Element

### OOP

* Instanzmethoden bekommen explizit `self` übergeben
* Zugriff auf Instanzvariablen nur über `self.var`
* Kein `private`, Konvention `self._private`

```python
class Person:
    
    def __init__(self, name, age)-> None:
        self._name = name
        self._age = age
        
    def is_of_age(self) -> bool:
        return self._age >= 18

person = Person('Name', 19)

print(person.is_of_age())
person._age = 17
person._age
```

* Python unterscheidet nicht zwischen Klassen und Interfaces
* Multiple Inheritance
* Es wird in allen Superklassen gesucht und der erste Hit gewinnt

```python
class Studier:
    
    def is_of_age(self) -> bool:
        return False

class Student(Studier, Person):
    
    def __init__(self, name, age, number)-> None:
        super().__init__(name, age)
        self._number = number
        
student = Student('Name', 39, 42)
student.is_of_age()
```
