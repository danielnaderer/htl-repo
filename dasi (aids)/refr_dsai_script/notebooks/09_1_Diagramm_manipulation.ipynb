# 📓 Diagramm-Manipulation Part 1:

## Daten
Es wurden synthetische Noten (n=40 je Klasse) erzeugt, die leicht variieren. Werte sind auf den Bereich 0–100 begrenzt. Die gleichen Daten liegen auch als CSV vor (`noten_4a_4b_4c.csv`).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Read CSV

💡 df.melt(): wandelt die Spalten (4A, 4B, 4C) in ein langes Format für Seaborn.

```python
df = pd.read_csv('../data/noten_4a_4b_4c.csv')
df_melt = df.melt(var_name="Klasse", value_name="Note (%)")
df.head()
```

# Welche Klasse ist richtig schlecht und um wieviel!

```python
plt.figure(figsize=(6, 4))
ax = sns.barplot(
    data=df_melt,
    x="Klasse",
    y="Note (%)",
    hue="Klasse",
    errorbar="se",
    palette="pastel",
    legend=False
)

# Titel und Achsen
ax.set_title("Durchschnittliche Noten pro Klasse (Seaborn ≥ 0.14-kompatibel)")
ax.set_xlabel("Klassen")
ax.set_ylabel("Noten (%)")
ax.set_ylim(70, 85)

plt.show()
```

```python
plt.figure(figsize=(6, 4))
ax = sns.barplot(
    data=df_melt,
    x="Klasse",
    y="Note (%)",
    hue="Klasse",        # 👈 notwendig, um palette korrekt zu verwenden
    errorbar="sd",       # 👈 zeigt Standardabweichung als Fehlerbalken (statt Konfidenzintervall).
    palette="pastel",    # 👈 sorgt für ein minimalistisches, helles Farbschema.
    legend=False         # 👈 blendet doppelte Legende aus
)

# Titel und Achsen
ax.set_title("Durchschnittliche Noten pro Klasse (Seaborn ≥ 0.14-kompatibel)")
ax.set_xlabel("Klassen")
ax.set_ylabel("Noten (%)")
ax.set_ylim(0, 100)

plt.show()
```
