# 🐧 Seaborn – Style

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Beispiel-Datensatz
df = sns.load_dataset('penguins')
df.head()
```

```python
sns.set_theme(style="whitegrid", context="talk", palette="colorblind")

# Scatterplot
sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species", style="sex", s=120)
sns.despine()
plt.title("Schnabelmaße nach Species & Sex")
plt.show()
```

```python

# Histogramm
sns.histplot(data=df, x="flipper_length_mm", hue="species", kde=True, element="step")  #poly, bars, step
#sns.despine()
plt.title("Flossenlängen – Verteilungen je Species")
plt.show()
```

```python

# Boxplot
order = df.groupby("species")["body_mass_g"].median().sort_values().index
sns.boxplot(data=df, x="species", y="body_mass_g", order=order)
sns.despine()
plt.title("Körpergewicht (Median-sortiert)")
plt.show()
```

```python

# Heatmap
cmap = sns.cubehelix_palette(start=.5, rot=-.5, as_cmap=True)
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap=cmap, annot=True, fmt=".2f")
sns.despine(left=True, bottom=True)
plt.title("Korrelationen (cubehelix)")
plt.show()
```
