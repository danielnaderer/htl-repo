```python
import numpy as np
import time

zahlen_liste = list(range(1_000_000))
zahlen_array = np.arange(1_000_000)

# 1. Liste
start = time.perf_counter()
ergebnis_liste = [x * 2 for x in zahlen_liste]
dauer_liste = time.perf_counter() - start
print(f"Liste: {dauer_liste * 1000:8.2f} ms")

# 2. Array
start = time.perf_counter()
ergebnis_array = zahlen_array * 2
dauer_array = time.perf_counter() - start
print(f"Array: {dauer_array * 1000:8.2f} ms")

# 3. Faktor
print(f"Das Array ist {dauer_liste / dauer_array:.0f}-mal schneller.")

# 4. Speicher
print(f"Speicher Array: {zahlen_array.nbytes / 1_000_000:.1f} MB")
```
