# Stellen wir die spärlich vorhanden Daten korrekt dar:

```python
import matplotlib.pyplot as plt
import pandas as pd

# -----------------------------
# Datensätze (geschätzt aus der Originalgrafik)
# -----------------------------
data = {
    "year": [1985, 1988, 1990, 1993, 1995, 1998, 2000, 2003, 2005, 2007, 2010, 2012, 2013],
    "gun_deaths": [870, 800, 750, 500, 550, 650, 600, 550, 450, 600, 700, 750, 721]
}

df = pd.DataFrame(data)

# -----------------------------
# Abbildung vorbereiten
# -----------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 5), constrained_layout=True)

# -----------------------------
# 1️⃣ "Falsche" (Original-)Darstellung: invertierte Y-Achse
# -----------------------------
axes[0].plot(df["year"], df["gun_deaths"], marker="o", color="firebrick", linewidth=2)
axes[0].axvline(2005, color="gray", linestyle="--", label="Stand Your Ground (2005)")
axes[0].invert_yaxis()  # ❗️wichtig: Achse umdrehen
axes[0].set_title("Originaldarstellung (invertierte Achse)")
axes[0].set_xlabel("Jahr")
axes[0].set_ylabel("Morde mit Schusswaffen")
axes[0].grid(True, linestyle="--", alpha=0.4)
axes[0].legend()

# -----------------------------
# 2️⃣ Korrekte Darstellung: normale Achse
# -----------------------------
axes[1].plot(df["year"], df["gun_deaths"], marker="o", color="firebrick", linewidth=2)
axes[1].axvline(2005, color="gray", linestyle="--", label="Stand Your Ground (2005)")
axes[1].set_title("Korrekte Darstellung (aufsteigende Achse)")
axes[1].set_xlabel("Jahr")
axes[1].set_ylabel("Morde mit Schusswaffen")
axes[1].grid(True, linestyle="--", alpha=0.4)
axes[1].legend()

plt.suptitle("Einfluss der Achsenrichtung auf die Wahrnehmung von Daten", fontsize=14)
plt.show()
```
