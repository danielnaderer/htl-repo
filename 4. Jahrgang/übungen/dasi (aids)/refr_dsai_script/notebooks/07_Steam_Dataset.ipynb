```python
import pandas as pd

df = pd.read_csv("../../refr_exercises/data/steamData/steam.csv")
# Bei sehr großen Datensätzen kannst du Teilmengen laden:
# df = pd.read_csv("steam.csv", nrows=100000)
print(df.head())
```

# Auswertung

```python
# Top 10 Spiele nach positiver Bewertung
df["positive_ratio"] = df["positive_ratings"] / (df["positive_ratings"] + df["negative_ratings"])
print(df.sort_values("positive_ratio", ascending=False).head(10))

print(df.columns)

# Durchschnittspreis pro Genre
print(df.groupby("genres")["price"].mean())
```
